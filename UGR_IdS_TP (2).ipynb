#### 1. Metodología de Adquisición de Datos para el Negocio
Para este proyecto, utilizaremos un dataset preexistente y validado que simula datos recolectados en el campo. En un entorno productivo real, la adquisición automatizada se estructuraría de la siguiente manera:
* **Datos de Suelo (N, P, K, pH):** Capturados mediante sensores IoT insertados en el suelo (por ejemplo, sensores ópticos o de conductividad eléctrica para NPK, y sensores de electrodos para pH) con transmisión vía LoRaWAN o redes celulares.
* **Datos Climáticos (Temperatura, Humedad, Precipitación):** Sincronizados en tiempo real mediante APIs de estaciones meteorológicas locales o servicios globales (como OpenWeatherMap o el Servicio Meteorológico Nacional) basados en la geolocalización del lote.

#### 2. Alternativa de Construcción del Set de Datos (Enfoque manual/laboratorio)
Si no dispusiéramos del dataset ya unificado, el proceso de construcción propia implicaría:
1. **Muestreo Físico de Suelos:** Recolección física de muestras de suelo en cuadrículas georreferenciadas dentro de diferentes fincas.
2. **Análisis de Laboratorio:** Envío de muestras a laboratorios de agronomía para cuantificar Nitrógeno (N), Fósforo (P), Potasio (K) en mg/kg (o ppm) y pH.
3. **Registro Histórico de Cultivos:** Entrevistar a los productores para registrar qué cultivos se sembraron históricamente en cada lote y si el rendimiento de la cosecha fue exitoso (etiquetado de datos).
4. **Asociación Climática:** Cruzar las coordenadas de los lotes con bases de datos climatológicas históricas mensuales de la región durante el ciclo de vida del cultivo.

#### 3. Identificación de Desafíos y Problemas de Calidad de Datos
* **Fallas de Sensores (Ruido y Outliers):** Los sensores de campo expuestos a la intemperie pueden enviar mediciones erráticas (ej. humedad del 150% o temperaturas negativas imposibles).
* **Datos Faltantes (Missing Values):** Pérdidas de conectividad en el campo que causen vacíos temporales en las series climáticas.
* **Desbalance de Clases:** Tener demasiados registros para un cultivo muy popular (ej. arroz) y muy pocos para otros (ej. lentejas), lo cual sesgaría el modelo de recomendación.
* **Descalibración del pH:** Los sensores de pH sufren un desgaste rápido por la acidez de la tierra, entregando lecturas desfasadas si no se calibran periódicamente.

### Desarrollo de la Fase #4: Modelado / Modeling

#### 1. Selección de la Variable Objetivo
La variable objetivo seleccionada para este estudio es **`label`** (tipo de cultivo recomendado).
* **Justificación:** Es una variable categórica nominal que contiene los diferentes tipos de cultivos óptimos (como arroz, maíz, café, etc.) según las características edafológicas y climáticas del suelo. Predecir esta variable de manera precisa cumple directamente con el objetivo general del negocio: recomendar el cultivo ideal para maximizar el rendimiento agrícola.

#### 2. Investigación del Algoritmo de Ciencia de Datos: Random Forest (Bosques Aleatorios)
Para resolver este problema de clasificación multiclase, hemos seleccionado el algoritmo **Random Forest Classifier**.

* **¿Qué es?:** Es un algoritmo de aprendizaje supervisado que combina múltiples árboles de decisión independientes (ensamble de tipo *Bagging*) para realizar una predicción robusta mediante votación mayoritaria.
* **¿Por qué es altamente útil para este problema?:**
  1. **Manejo de No-Linealidad:** La relación entre las variables de entrada (como la interacción entre el nitrógeno, el pH y el clima) y el tipo de cultivo óptimo no es lineal. Los árboles de decisión capturan estas interacciones complejas de forma natural.
  2. **Robustez ante el Ruido y el Sobreajuste:** Al promediar las predicciones de muchos árboles entrenados con diferentes subconjuntos de datos, Random Forest reduce drásticamente el riesgo de sobreajuste (*overfitting*) en comparación con un único árbol de decisión.
  3. **Importancia de las Variables:** Nos permite calcular y entender qué variables (por ejemplo, el pH o la cantidad de lluvia) influyen más en la recomendación de cada cultivo, aportando explicabilidad al negocio agrícola.

````python
# Importación de librerías de machine learning para la Fase #4
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

print("Librerías de modelado (scikit-learn) cargadas correctamente.")
````

````python
# 1. Definición de variables predictoras (X) y objetivo (y)
X = df.drop(columns=['label'])
y = df['label']

# 2. División en conjuntos de Entrenamiento (Train) y Prueba (Test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)

# 3. Inicialización y entrenamiento del modelo
# Elegimos Random Forest por su solidez ante datos tabulares multidimensionales
modelo_cultivos = RandomForestClassifier(n_estimators=100, random_state=42)
modelo_cultivos.fit(X_train, y_train)

print("Modelo Random Forest entrenado exitosamente con los datos de entrenamiento.")
print(f"Tamaño del set de entrenamiento: {X_train.shape[0]} muestras")
print(f"Tamaño del set de testeo: {X_test.shape[0]} muestras")
````

### Desarrollo de la Fase #5: Evaluación / Evaluation

#### 1. Explicación de las Métricas de Evaluación Seleccionadas
Para medir el desempeño de nuestro recomendador multiclase, utilizamos cuatro métricas fundamentales de la matriz de confusión:

* **Exactitud (Accuracy):** Mide la proporción de predicciones correctas sobre el total de casos evaluados. Es útil como métrica general, pero puede ser engañosa si las clases estuvieran muy desbalanceadas.
* **Precisión (Precision):** De todos los cultivos que el modelo predijo como una clase específica (ej. *maíz*), ¿cuántos eran realmente correctos? Minimiza los falsos positivos (recomendar erróneamente un cultivo no apto).
* **Sensibilidad / Recuperación (Recall):** De todos los cultivos de una clase que realmente estaban presentes en el set de prueba, ¿cuántos logró detectar el modelo? Minimiza los falsos negativos.
* **F1-Score:** El promedio armónico entre la Precisión y el Recall. Proporciona una métrica única y balanceada sobre el desempeño del modelo para cada cultivo.

````python
# Importación de librerías para evaluación del modelo
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
import seaborn as sns

print("Librerías de evaluación y métricas cargadas correctamente.")
````

````python
# 1. Realizar predicciones con el conjunto de prueba (Test)
y_pred = modelo_cultivos.predict(X_test)

# 2. Calcular la precisión general
exactitud = accuracy_score(y_test, y_pred)
print(f"=== Exactitud del Modelo (Accuracy): {exactitud:.4f} ===\n")

# 3. Mostrar el reporte detallado de clasificación
print("=== Reporte de Clasificación ===")
print(classification_report(y_test, y_pred))

# 4. Generar y graficar la Matriz de Confusión
fig, ax = plt.subplots(figsize=(12, 10))
cm = confusion_matrix(y_test, y_pred, labels=modelo_cultivos.classes_)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=modelo_cultivos.classes_)
disp.plot(ax=ax, cmap='YlGnBu', xticks_rotation='vertical')
plt.title('Matriz de Confusión - Recomendación de Cultivos')
plt.tight_layout()
plt.show()
````

### Validación Cruzada y Análisis de la Confianza

Para verificar que el 99.55% de exactitud no sea fruto de una división afortunada train/test, se realiza una **validación cruzada de 5 pliegues** sobre todo el dataset. Además, se analiza la **confianza** de cada predicción (máxima probabilidad asignada por el clasificador), que es exactamente la métrica que la aplicación Streamlit muestra al usuario final.

> **Nota de interpretación:** la confianza es una medida *por predicción individual* y **no equivale** a la exactitud promedio. En zonas del espacio de features sin muestras de entrenamiento (combinaciones arbitrarias de los sliders de la app), la confianza cae, aunque el modelo acierte casi siempre sobre datos similares a los del entrenamiento.

````python
from sklearn.model_selection import cross_val_score

# 1. Validación cruzada de 5 pliegues con el mismo clasificador
cv_scores = cross_val_score(RandomForestClassifier(n_estimators=100, random_state=42),
                            X, y, cv=5, scoring='accuracy')

print("=== VALIDACIÓN CRUZADA DE 5 PLIEGUES (Accuracy) ===")
print(f"Pliegues: {[f'{s:.4f}' for s in cv_scores]}")
print(f"Accuracy promedio: {cv_scores.mean():.4f} ± {cv_scores.std():.4f}")
print(f"Mín: {cv_scores.min():.4f} | Máx: {cv_scores.max():.4f}\n")

# 2. Confianza (máxima probabilidad) de cada predicción en el conjunto de prueba
conf_test = modelo_cultivos.predict_proba(X_test).max(axis=1)

print("Distribución de la confianza en el test (440 muestras):")
print(f"  media={conf_test.mean():.3f} | mediana={np.median(conf_test):.3f} | mín={conf_test.min():.3f} | máx={conf_test.max():.3f}")
print(f"  % de muestras con confianza < 0.70: {(conf_test < 0.70).mean()*100:.1f}%\n")

# 3. Histograma de la confianza en el conjunto de prueba
plt.figure(figsize=(9, 4.5))
sns.histplot(conf_test, bins=25, color='steelblue', edgecolor='white')
plt.axvline(np.median(conf_test), color='crimson', linestyle='--', linewidth=1.5,
            label=f'Mediana = {np.median(conf_test):.2f}')
plt.title('Distribución de la Confianza del Modelo (máx. probabilidad) en el Conjunto de Prueba')
plt.xlabel('Confianza (max predict_proba)')
plt.ylabel('Cantidad de muestras')
plt.legend()
plt.tight_layout()
plt.show()
````

### Interpretación de la Matriz de Confusión (Datos Reales)

Con un **99.55% de exactitud (Accuracy)** en el conjunto de prueba, la matriz de confusión revela un desempeño casi perfecto del clasificador **Random Forest**:

1. **Diagonal Principal Sólida:** Prácticamente la totalidad de las 20 muestras de prueba por cada cultivo (como *apple*, *banana*, *orange*, etc.) se sitúan en la diagonal, indicando que el modelo predijo exactamente el cultivo correcto.
2. **Confusiones Mínimas Identificadas:**
   * **Jute vs. Rice:** Hubo una leve confusión donde 1 muestra de arroz (*rice*) fue clasificada erróneamente como yute (*jute*), y viceversa. Esto es biológicamente coherente, ya que ambos cultivos requieren condiciones de humedad y precipitación extremadamente altas y muy similares.
   * **Blackgram vs. Maize:** Se observa una confusión mínima (1 muestra) entre *blackgram* (legumbre) y *maize* (maíz), lo que sugiere rangos de nutrientes de suelo muy cercanos en ese punto del espacio muestral.
3. **Validación Cruzada (5 pliegues):** el desempeño se confirma más allá de una única división train/test: accuracy promedio de **99.45% ± 0.34%** (pliegues entre 98.86% y 99.77%), es decir, un error medio ≈ **0.55%**.
4. **Conclusión de Negocio — alcance y límites:** el modelo es **altamente confiable para ser integrado en la interfaz del usuario cuando los parámetros ingresados son similares a los datos de entrenamiento** (exactitud ≈ 99.5%; el error real en test fue 2/440 = 0.45%, con intervalo de confianza al 95% ≈ 0.05%–1.6%). Este porcentaje es un **promedio de desempeño, no una garantía por predicción individual**: la confianza (máxima probabilidad) tiene mediana 0.99 en datos similares a los del test, pero **cae por debajo del 50% en combinaciones de parámetros no representadas en el dataset** (zonas sin muestras de entrenamiento).
